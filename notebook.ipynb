## 1. Python Basics

```python
# Built-in types
name   = "David"
age    = 30
height = 1.80
active = True

print(f"Name: {name}, Age: {age}, Height: {height}m, Active: {active}")
print(f"Types: {type(name)}, {type(age)}, {type(height)}, {type(active)}")
```

```python
# Collections
skills     = ["HTML", "CSS", "JavaScript", "PHP", "Python"]   # list — ordered, mutable
dimensions = (1920, 1080)                                       # tuple — ordered, immutable
tools      = {"editor": "VS Code", "vcs": "Git", "os": "Linux"} # dict — key/value
unique_tags = {"frontend", "backend", "design"}                  # set — unordered, unique

print("Skills:", skills)
print("First skill:", skills[0], "| Last:", skills[-1])
print("Sliced:", skills[1:3])
print("Editor:", tools["editor"])
```

## 2. Control Flow

```python
# if / elif / else
experience_years = 5

if experience_years < 2:
    level = "Junior"
elif experience_years < 6:
    level = "Mid-level"
else:
    level = "Senior"

print(f"{experience_years} years → {level} developer")

# for loop + enumerate
print("\nSkill list:")
for i, skill in enumerate(skills, start=1):
    print(f"  {i}. {skill}")

# while loop
count = 0
while count < 3:
    print(f"  Loop iteration {count}")
    count += 1
```

## 3. Functions

```python
def greet(name: str, language: str = "fr") -> str:
    """Return a greeting in the requested language."""
    messages = {
        "fr": f"Bonjour, {name} !",
        "en": f"Hello, {name}!",
        "es": f"¡Hola, {name}!",
    }
    return messages.get(language, f"Hi, {name}!")

print(greet("David"))
print(greet("David", "en"))
print(greet("David", "es"))

# Lambda (anonymous function)
square = lambda x: x ** 2
print("\nSquares:", list(map(square, range(1, 6))))
```

## 4. List / Dict Comprehensions

```python
numbers = range(1, 11)

# List comprehension
evens   = [n for n in numbers if n % 2 == 0]
squares = [n ** 2 for n in numbers]

# Dict comprehension
square_map = {n: n ** 2 for n in numbers}

print("Evens:", evens)
print("Squares:", squares)
print("Square map:", square_map)
```

## 5. Classes & OOP

```python
class Developer:
    species = "Human"

    def __init__(self, name: str, skills: list[str]):
        self.name   = name
        self.skills = skills

    def introduce(self) -> str:
        return f"I'm {self.name} and I work with: {', '.join(self.skills)}."

    def add_skill(self, skill: str):
        self.skills.append(skill)

    def __repr__(self):
        return f"Developer(name={self.name!r}, skills={self.skills!r})"


class FrontendDev(Developer):
    def __init__(self, name, skills, frameworks):
        super().__init__(name, skills)
        self.frameworks = frameworks

    def introduce(self):
        base = super().introduce()
        return f"{base} Frameworks: {', '.join(self.frameworks)}."


david = FrontendDev(
    name="David",
    skills=["HTML", "CSS", "JavaScript"],
    frameworks=["React", "Vue"]
)
print(david.introduce())
david.add_skill("Python")
print(david)
```

## 6. Useful Standard Library Modules

```python
import os
import datetime
import json
import random

# datetime
now = datetime.datetime.now()
print("Now:", now.strftime("%Y-%m-%d %H:%M:%S"))

# json
profile = {"name": "David", "role": "Frontend Developer", "skills": skills}
json_str = json.dumps(profile, ensure_ascii=False, indent=2)
print("\nJSON output:\n", json_str)
parsed  = json.loads(json_str)
print("Parsed role:", parsed["role"])

# random
print("\nRandom skill pick:", random.choice(skills))
sample = random.sample(skills, k=3)
print("Random sample:", sample)
```

## 7. NumPy — Numerical Computing

```python
import numpy as np

arr = np.array([1, 2, 3, 4, 5])
print("Array:", arr)
print("Mean:", arr.mean(), "| Std:", arr.std(), "| Sum:", arr.sum())

matrix = np.arange(9).reshape(3, 3)
print("\nMatrix:\n", matrix)
print("Transposed:\n", matrix.T)

# Broadcasting
print("\nArray * 2:", arr * 2)
print("Array ** 2:", arr ** 2)
```

## 8. Pandas — Data Manipulation

```python
import pandas as pd

data = {
    "skill":      ["HTML", "CSS", "JavaScript", "PHP", "Python"],
    "years":      [10, 10, 8, 5, 2],
    "proficiency": ["Expert", "Expert", "Advanced", "Intermediate", "Beginner"],
}

df = pd.DataFrame(data)
print(df)
print("\nDescriptive stats:")
print(df["years"].describe())

# Filter
print("\nSkills with > 5 years:")
print(df[df["years"] > 5])
```

## 9. Matplotlib — Data Visualization

```python
import matplotlib.pyplot as plt

fig, axes = plt.subplots(1, 2, figsize=(12, 4))

# Bar chart — skill years
axes[0].bar(df["skill"], df["years"], color=["#E20613", "#2BA9E0", "#36609F", "#066C9A", "#F5A623"])
axes[0].set_title("Years per Skill")
axes[0].set_xlabel("Skill")
axes[0].set_ylabel("Years")

# Pie chart — distribution
axes[1].pie(df["years"], labels=df["skill"], autopct="%1.0f%%", startangle=90)
axes[1].set_title("Skill Distribution")

plt.tight_layout()
plt.show()
```
